# 패션 MNIST

Fashion MNIST는 10개의 클래스로 이루어진 옷과 신발 이미지를 모아둔 데이터셋이다.

- 28×28 픽셀
- 흑백 1채널 이미지
- 70,000(훈련 60000, 테스트 10000)장
- 라벨은 0~9
  | 라벨 | 의미 |
  | --- | --- |
  | 0 | T-shirt/top |
  | 1 | Trouser |
  | 2 | Pullover |
  | 3 | Dress |
  | 4 | Coat |
  | 5 | Sandal |
  | 6 | Shirt |
  | 7 | Sneker |
  | 8 | Bag |
  | 9 | Ankle boot |
- 예:

  ![](https://miro.medium.com/v2/resize:fit:1400/format:webp/1*1zuF9G8fq3hUTafkBLcRkQ.png)

## 데이터셋 다운로드

```python
### Scikit-learn으로 다운 받기
```

```python
from sklearn.datasets import fetch_openml

fashion_mnist = fetch_openml("Fashion-MNIST", version=1, as_frame=False)
fashion_mnist.keys()
```

```python
print(fashion_mnist["DESCR"])
```

## Matplotlib 으로 살펴보기

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(5,5))
plt.imshow(fashion_mnist.data[0].reshape(28, 28), cmap='gray')
plt.axis('off')
plt.show()

print(fashion_mnist.data[0])
print("이 숫자는 ", fashion_mnist.target[0], "입니다.")
```
