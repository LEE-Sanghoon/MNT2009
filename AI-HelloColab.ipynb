### 파이썬 코드를 여러 셀로 나눠 실행하기

1. 초기화 내용과 작업 순서를 지시하는 코드를 서로 다른 셀로 작성한다.
2. 각각의 셀을 순서대로 "실행" 한다.

```python
import time

start = time.time()
sum = 0
```

```python
for i in range(1, 10000001):
  sum = sum + i

end = time.time()

print('1+2+...+10000001=', sum)
print('소요시간은 ', end-start, '초 입니다')
```

### TTS (Text-to-Speech): 음성출력하기

1. 음성출력생성 모듈("gTTS")을 설치
2. 음성 파일 작성
3. 작성된 음성 파일 재생
4. 음성 파일은 어디에 저장되었는지 확인

```python
!pip3 install gTTS
```

```python
from gtts import gTTS

text_to_say = "안녕하세요. 인공지능 수업입니다."
language = "ko"

gtts_object = gTTS(text = text_to_say,
                   lang = language,
                   slow = False)

gtts_object.save('/content/gtts.wav')
```
