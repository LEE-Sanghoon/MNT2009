- XOR 게이트 예제

```python
import os
os.environ["KERAS_BACKEND"] = "torch"

import keras
from keras import layers
import numpy as np


x_train = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
], dtype="float32")

y_train = np.array([
    [0],
    [1],
    [1],
    [0]
], dtype="float32")


model = keras.Sequential([
    layers.Input(shape=(2,)),
    layers.Dense(4, activation="sigmoid"),
    layers.Dense(1, activation="sigmoid")
])

model.compile(
    optimizer=keras.optimizers.SGD(learning_rate=0.5),
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

model.fit(x_train, y_train, epochs=5000, verbose=0)

predictions = model.predict(x_train)

print("\nXOR 테스트 결과")
print("-" * 30)

for x, y, pred in zip(x_train, y_train, predictions):
    result = 1 if pred[0] >= 0.5 else 0
    print(
        f"입력: {x.tolist()}, "
        f"정답: {int(y[0])}, "
        f"예측값: {pred[0]:.4f}, "
        f"분류 결과: {result}"
    )
```
