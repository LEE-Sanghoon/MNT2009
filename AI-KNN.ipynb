# K-최근접 이웃(KNN, K-Nearest Neighbor) 알고리즘

* 새 데이터와 가장 가까운 기존 데이터 k를 보고 예측하는 단순한 지도학습 알고리즘
  - 새로운 데이터가 입력되면,
  - 기존 데이터들과의 거리를 계산하고,
  - 가장 가까운 k개 이웃을 찾은 뒤,
  - 그 이웃들의 다수결이나 평균으로 예측 한다.

* 장점:
  - 이해하기 쉽고 구현이 간단함
  - 복잡한 학습 과정이 거의 없음
  - 기본 성능이 좋은 편

* 단점:
  - 데이터가 증가하면, 예측 소요 시간이 증가
  - 거리 기반이므로 스케일링의 영향을 많이 받음
  - k값 선택에 따라 성능이 달라짐
  - 차원이 높으면 성능이 떨어질 수 있음

### KNN 분류기 학습

* 학습 지원 모듈(scikit-learn) 설치
* 학습 데이터 생성
* 모델 학습기 생성 (KNeighborsClassifier)
* 학습 진행 (" .fit( *train_data*, *train_target* ) ")
* 학습 평가 (" .score( *test_data*, *test_target* ) ")
* 예측 수행 (" .predict( *input_data* )

* 전체 데이터를 학습에 사용

```python
length = bream_length + smelt_length
weight = bream_weight + smelt_weight

fish_data = [ [l, w] for l, w in zip(length, weight)]
fish_target = [1]*35 + [0]*14
```

```python
from sklearn.neighbors import KNeighborsClassifier
kn = KNeighborsClassifier()
kn.fit(fish_data, fish_target)
kn.score(fish_data, fish_target)
```

```python
kn.predict( [[30, 600]])
```

* 학습 데이터(Training Dataset)와 검증 데이터(Validation Dataset) 구분

```python
train_input = fish_data[:35]
train_target = fish_target[:35]
test_input = fish_data[35:]
test_target = fish_target[35:]
```

```python
from sklearn.neighbors import KNeighborsClassifier
kn = KNeighborsClassifier()
kn = kn.fit(train_input, train_target)
kn.score(test_input, test_target)
```

```python
import numpy as np
input_arr = np.array(fish_data)
target_arr = np.array(fish_target)
index = np.arange(49)
np.random.shuffle(index)
print(index)
```

```python
train_input = input_arr[index[:35]]
train_target = target_arr[index[:35]]
test_input = input_arr[index[35:]]
test_target = target_arr[index[35:]]
```

```python
kn = kn.fit(train_input, train_target)
kn.score(test_input, test_target)
```
