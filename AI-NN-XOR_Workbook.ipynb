# 인공신경망 자율학습지

- XOR 문제
  
    | x1 | x2 | y |
    | --- | --- | --- |
    | 0 | 0 | 0 |
    | 0 | 1 | 1 |
    | 1 | 0 | 1 |
    | 0 | 0 | 0 |

- 네트워크 구조  


```python
import numpy as np

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def sigmoid_derivative(a):
    return a * (1 - a)

# XOR 데이터
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([
    [0],
    [1],
    [1],
    [0]
])

# 가중치 초기화
W1 = np.array([
    [0.1, 0.3],
    [0.2, 0.4]
])

b1 = np.array([[0.1, 0.1]])

W2 = np.array([
    [0.5],
    [0.6]
])

b2 = np.array([[0.1]])

lr = 0.1

for epoch in range(10000):
    # 순전파
    z1 = X @ W1 + b1
    h = sigmoid(z1)

    z2 = h @ W2 + b2
    y_hat = sigmoid(z2)

    loss = np.mean(0.5 * (y_hat - y) ** 2)

    # 역전파
    delta2 = (y_hat - y) * sigmoid_derivative(y_hat)

    dW2 = h.T @ delta2 / len(X)
    db2 = np.mean(delta2, axis=0, keepdims=True)

    delta1 = (delta2 @ W2.T) * sigmoid_derivative(h)

    dW1 = X.T @ delta1 / len(X)
    db1 = np.mean(delta1, axis=0, keepdims=True)

    # 가중치 갱신
    W2 -= lr * dW2
    b2 -= lr * db2
    W1 -= lr * dW1
    b1 -= lr * db1

    if epoch % 1000 == 0:
        print(epoch, loss)

print("예측 결과")
print(y_hat)
```
