# IRIS 데이터셋

IRIS (분꽃)은 다음의 3가지 종류(Setosa, Versicolor, Virginica)가 있다.

![](https://miro.medium.com/v2/resize:fit:600/format:webp/1*ZDZcu5VdTGdl9-CguPLiSg.png)

사이키런에는 머신 러닝을 쉽게 배울 수 있도록 샘플 데이터셋들을 가지고 있다. 그 중에서 Iris 데이터셋도 있다. Iris 데이터셋은 자료형이 딕셔너리 다운 로드 받은 후 Key 값을 먼저 확인 한다.

## 데이터셋 다운로드

```python
from sklearn import datasets

iris = datasets.load_iris()
iris.keys()
```

Key에 있는 `DESCR`을 사용해 데이터셋의 정보를 확인한다.

```python
print(iris.DESCR)
```

데이터셋 정보를 확인 할 수 있다.
- 150개의 샘플 데이터
- 4개의 특징(feature) 정보
  - sepal(꽂받침)의 length/width(길이/너비), petal(꽃잎)의 length/width
- 3개의 품종
  - Iris-Setosa, Iris-Versicolour, Iris-Virginica

코드를 통해서 확인 할 수도 있다.

```python
print(iris.feature_names)
```

```python
print(iris.target_names)
```

샘플을 순서대로 출력

```python
for i in range(0, len(iris.data)):  # 샘플을 순서대로 출력
    print( i+1, iris.data[i], iris.target[i])
```

## Pandas로 데이터셋 살펴보기

판다스의 DataFrame은 데이터를 분석하고 사용하기 편리 하다.

```python
import pandas as pd

df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['species'] = iris.target
```

처음 몇 (10) 줄 살펴 보기

```python
df.head(10)
```

```python
df.info()
```

### 데이터 확인하기

통계 데이터 확인

```python
df.describe()
```

중복 데이터 확인

```python
print(df.duplicated().sum())

df.loc[df.duplicated(), :]
```

## Seaborn 으로 데이터 분석하기

```python
import seaborn as sns

sns.pairplot(df, hue='species')
```

히스토그램

```python
sns.displot(df, x='sepal_length', hue='species', kind='hist')
```

히트맵

```python
sns.heatmap(data=df.select_dtypes(include='number').corr(), cbar=True, annot=True)
```

## plotly 

- Colab에서 출력
  `fig.show(renderer='colab')` 으로 수정

```python
import plotly.express as px

df = px.data.iris()
fig = px.scatter_3d(df, 
                    x ='sepal_length',
                    y = 'sepal_width',
                    z = 'petal_width',
                    color = 'species')

fig.show()
```
