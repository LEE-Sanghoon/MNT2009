# 손글씨 숫자 이미지 데이터셋

MNIST(Modified National Institute of Standards and Technology)는 0부터 9까지의 손글씨 숫자 이미지를 모아둔 데이터셋이다.

각 이미지는 다음과 같은 형태이다.
- 28 $\times$ 28
- 흑백 이미지
- 숫자 하나가 중앙에 위치
- 라벨은 0~9 중 하나
- 예:

    ![](https://miro.medium.com/v2/resize:fit:600/format:webp/1*84PejJCKcc1dXH1gluzr_Q.png)

## 데이터셋 다운로드

### Scikit-learn 으로 다운 받기

- Download Full MNIST (70,000 images, 28x28 pixels) 

```python
from sklearn import datasets

# Download the MNIST dataset
# 'as_frame=False' returns NumPy arrays; 'True' returns a Pandas DataFrame
mnist = datasets.fetch_openml('mnist_784', version=1, as_frame=False)
mnist.keys()
```

```python
print(mnist.DESCR)
```

- Toy Digits Dataset (1,797 images, 8x8 pixels)

```python
# Toy Digits Dataset (1,797 images, 8x8 pixels)
from sklearn import datasets

mnist = datasets.load_digits()
mnist.keys()
```

```python
print(mnist.DESCR)
```

## 데이터 시각화

## Matplotlib 으로 살펴보기

- `fetch_openml (..., as_frame=False)` 로 받은 데이터셋의 경우

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(5,5))
plt.imshow(mnist.data[0].reshape(28, 28), cmap='gray')
plt.axis('off')
plt.show()

print(mnist.data[0])
print("이 숫자는 ", mnist.target[0], "입니다.")
```

- `fetch_openml (..., as_frame=True)` 로 받은 데이터셋의 경우
  
  Pandas의 DataFrame 자료형에 경우 `.iloc[0].values`를 사용한다.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(5,5))
plt.imshow(mnist.iloc[0].values.reshape(28, 28), cmap='gray')
plt.axis('off')
plt.show()

print(mnist.data[0])
print("이 숫자는 ", mnist.target[0], "입니다.")
```
